```python
s = input("Enter a string: ")
rev = ""

for i in s:
    rev = i + rev

print("Reversed string:", rev)
```

```python

a = [1, 2, 2, 3, 4, 3, 5]
b = []

for i in a:
    if i not in b:
        b.append(i)

print(b)
```

```python

n = int(input("Enter a 3-digit number: "))

a = n % 10
b = (n // 10) % 10
c = n // 100

rev = a * 100 + b * 10 + c

print("Reversed number:", rev)
```

```python

class Student:
    def __init__(self):
        self.rollno = int(input("Enter roll number: "))
        self.name = input("Enter student name: ")

        self.m1 = float(input("Enter marks of paper 1: "))
        self.m2 = float(input("Enter marks of paper 2: "))
        self.m3 = float(input("Enter marks of paper 3: "))
        self.m4 = float(input("Enter marks of paper 4: "))
        self.m5 = float(input("Enter marks of paper 5: "))

    def calculate(self):
        total = self.m1 + self.m2 + self.m3 + self.m4 + self.m5
        percentage = total / 5

        print("\nRoll Number:", self.rollno)
        print("Name:", self.name)
        print("Total Marks:", total)
        print("Percentage:", percentage, "%")

        if percentage >= 85:
            print("Student can take admission")
        else:
            print("Student cannot take admission")


s = Student()
s.calculate()

```

```python

class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(B):
  def showC(self):
    print("showC function")

obj = B()
obj.showA()
obj.showB()
obj.showC()
```

```python

class A:
    def show(self):
        print("Class A")

class B(A):
    def display(self):
        print("Class B")

class C(A):
    def display2(self):
        print("Class C")

class D(B, C):
    def show2(self):
        print("Class D")

obj = D()
obj.show()
obj.display()
obj.display2()
obj.show2()
```
