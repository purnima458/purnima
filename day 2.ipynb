```python
fruit_list1 = ['Apple', 'Berry', 'Cherry', 'Papaya']
fruit_list2 = fruit_list1
fruit_list3 = fruit_list1[:]

fruit_list2[0] = 'Guava'
fruit_list3[1] = 'Kiwi'

sum = 0

for ls in (fruit_list1, fruit_list2, fruit_list3):
    sum += 1
    if ls[0] == 'Guava':
        if ls[1] == 'Kiwi':
            sum += 20

print(sum)
```

```python
data = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]

def fun(m):
    v = m[0][0]

    for row in m:
        for element in row:
            if v < element:
                v = element

    return v

print(fun(data[0]))
```

```python
# with statement

with open("myfile.txt", 'w') as f:
    f.write("Python\n")
    f.write("is\n")
    f.write("easy\n")
    f.write("language\n")
    print("is file closed ?", f.closed)

print("Now is file closed ?", f.closed)
```

```python
# We can use seek() function to move cursor (file pointer) to
# specified location.

data = "Learning python is very easy for Ashish sir"

f = open("myfile.txt", 'w')
f.write(data)

with open("myfile.txt", 'r+') as f:
    fileData = f.read()
    print(fileData)

    print("Current Cursor Position is", f.tell())

    f.seek(15)
    print("Current Cursor Position is", f.tell())

    f.write("-JAVA-")

    f.seek(0)
    fileData = f.read()
    print(fileData)
```

```python


import os, sys

fname = input("Enter file name : ")

if os.path.isfile(fname):
    print("File exist :", fname)
    f = open(fname, 'r')
else:
    print("File is not exist :", fname)
    sys.exit(0)

lcount = wcount = ccount = 0

for line in f:
    lcount = lcount + 1
    ccount = ccount + len(line)
    words = line.split()
    wcount = wcount + len(words)

print("Total lines :", lcount)
print("Total words :", wcount)
print("Total characters :", ccount)

f.close()
```
