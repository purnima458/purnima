```python
import numpy as np

students = np.array([
    [85, 90, 78],  # Student 1
    [60, 55, 70],  # Student 2
    [95, 92, 88],  # Student 3
    [40, 50, 45],  # Student 4
    [72, 68, 75]   # Student 5
])

students
```

```python
avg_marks = np.mean(students, axis=1)

print("Average Marks :", avg_marks)
```

```python
failed_student=np.where(avg_marks<50)

print("Failed Students :", failed_student)

```

```python
# Performance Report

for i, avg in enumerate(avg_marks):
    status = "Pass" if avg > 50 else "Fail"
    print(f"Student {i+1}: Avg = {avg:.2f} | {status}")
```

```python
import pandas as pd

# Create data frame and series

data = {
    'Name': ['Prashant', 'Ashish', 'Sandip'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Paris', 'London']
}

df = pd.DataFrame(data)

print(df)
```

```python
# Bar Chart

months = ['Jan', 'Feb', 'Mar', 'April']
sales = [250, 300, 400, 350]

plt.bar(months, sales, color='skyblue')
plt.title('Monthly sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
plt.barh(months, sales, color='orange')
plt.title('Monthly sales (Horizontal)')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```

```python
# Scatter Plot (Comparing Height vs Weight)

height = [150, 160, 170, 180]
weight = [50, 60, 65, 75]

plt.scatter(height, weight, color='pink')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')

plt.show()
```

```python
# Pie Chart (Market Share)

import matplotlib.pyplot as plt

brands = ['Brand A', 'Brand B', 'Brand C']
shares = [45, 30, 25]

plt.pie(shares, labels=brands, autopct='%1.1f%%', startangle=90)

plt.title('Market Shares')
plt.axis('equal')
plt.show()
```
